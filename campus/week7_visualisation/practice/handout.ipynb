### line plot

```python
from matplotlib import pyplot as plt
import numpy as np
import pandas as pd
```

```python
years = [1980, 1990, 2000, 2010, 2016]
life_exp_men = [70.81, 73.16, 75.61, 78.81, 79.25]
life_exp_women = [76.81, 78.7, 80.36, 78.71, 82.93]

df = pd.DataFrame({'Year': years,
                   'Male': life_exp_men,
                   'Female': life_exp_women})

# create the plot
plt.plot(df['Year'], df['Male'], 'sg--' ,label='Male')
plt.plot(df['Year'], df['Female'], 'hb:', label='Female')

# 所以这里的 ‘’l里是 'marker, color, linestyle'
plt.xlabel('Year')
plt.ylabel('Life')
plt.title('UK life Expectancy')

plt.legend()
plt.grid(True)
plt.show()
```
