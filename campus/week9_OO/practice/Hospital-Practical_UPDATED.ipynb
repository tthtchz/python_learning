**Task 1: Patient Class**

Attributes:

* name: The patient's full name (e.g., "Alice Smith").
* date_of_birth: The patient's date of birth as a datetime.date object.
* weight: The patient's weight in kilograms.
* height: The patient's height in meters.
* sex: The patient's sex ("Male" or "Female").

Methods:

* `__init__(self, name, date_of_birth, weight, height, sex)`: Initialize the patient with the given attributes.

* `calculate_bmi()`: Calculate and return the BMI using the formula:  
$$\text{BMI} = \frac{\text{weight}}{\text{height}^2}$$

* `age()`: Calculate and return the patient’s age in years.
* `get_details()`: Return a formatted string with the patient's details, including name, age, weight, height, and BMI.
